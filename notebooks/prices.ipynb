# Plotting commodity prices

This notebook shows how to plot commodity prices, using an example model.

We first load the prices from the `commodity_prices.csv` output file.

```python
import pandas as pd

from muse2_data_analysis.helpers import get_example_output_dir

OUTPUT_DIR = get_example_output_dir()
prices = pd.read_csv(OUTPUT_DIR / "commodity_prices.csv")
prices.head()
```

```python
from matplotlib import pyplot as plt

COMMODITY_ID = "heat"

# This won't work if there are multiple regions
assert prices["region_id"].nunique() == 1

_, ax = plt.subplots()

prices[prices["commodity_id"] == COMMODITY_ID].pivot(
    index="milestone_year", columns="time_slice", values="price"
).plot(kind="bar", ax=ax)

ax.set_xlabel("Milestone year")
ax.set_ylabel("Price")
ax.legend(
    title="Time slice", bbox_to_anchor=(1.05, 1), loc="upper left", borderaxespad=0.0
);
```
